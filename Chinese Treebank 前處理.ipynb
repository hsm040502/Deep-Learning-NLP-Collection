在 NLTK 套件中，有個由中央研究院製作的 Chinese Treebank 語料庫，請幫我撰寫一段能夠載入該語料庫並印出原始文字的 Python 原始碼。

該程式碼的步驟如下所示：

*   引入下列函式庫： from nltk.corpus import sinica_treebank。
*   將函數庫中的斷詞重新組合成原始句子，完成後將其全部印出

撰寫此一程式時，請遵守下列限制：

*   將該程式碼撰寫成適合在 Google Colab 環境下執行的樣子。

```python
# -*- coding: utf-8 -*-
# 在 Google Colab 中執行此程式碼

import nltk
from nltk.corpus import sinica_treebank
import ssl # 用於處理可能的 SSL 憑證問題

# 嘗試處理 SSL 憑證問題 (在某些環境下需要)
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

# 下載 Sinica Treebank 語料庫 (如果尚未下載)
try:
    # 嘗試訪問語料庫以觸發下載檢查
    print("檢查 Sinica Treebank 語料庫是否存在...")
    _ = sinica_treebank.sents()
    print("語料庫已存在或載入成功。")
except LookupError:
    print("Sinica Treebank 語料庫未找到，開始下載...")
    nltk.download('sinica_treebank')
    print("語料庫下載完成。")
except Exception as e:
    print(f"載入或下載語料庫時發生錯誤: {e}")
    # 如果下載/載入失敗，則停止執行
    exit()

print("\n" + "="*30)
print("開始處理 Sinica Treebank 語料庫")
print("="*30 + "\n")

# 取得所有已斷詞的句子 (每個句子是一個詞的列表)
segmented_sentences = sinica_treebank.sents()

print(f"語料庫總句子數: {len(segmented_sentences)}\n")

print("將斷詞重新組合成原始句子並印出：")
print("-" * 40)

# 迭代每個已斷詞的句子
count = 0
for sentence_tokens in segmented_sentences:
    # 使用空字串將列表中的詞語連接起來，還原成未斷詞的句子
    original_sentence = "".join(sentence_tokens)
    print(original_sentence)
    count += 1
    # # 如果語料庫很大，可以取消下面幾行的註解，只印出前幾句作為範例
    # if count >= 20:
    #     print("\n(僅顯示前 20 句)")
    #     break

print("-" * 40)
print(f"\n已成功處理並印出 {count} 個句子。")
```

非常好，請幫我承接上面，撰寫一個能夠針對中文進行「斷詞」與「文本清理」的 Python 原始碼。

該程式碼的步驟如下所示：

*   請對上面產生的original_sentence以 jieba 套件進行「斷詞」，並執行「標點符號」移除的工作。
*   完成後將語料庫內建的斷詞結果nltk.corpus.sinica_treebank.sents()，與 jieba 斷詞結果並列，一次一句地全數印出。

撰寫此一程式時，請遵守下列限制：

*   將該程式碼撰寫成適合在 Google Colab 環境下執行的樣子。

```python
# 1. 安裝 jieba (如果尚未安裝)
#    使用 ! 可以在 Colab 中執行 shell 命令
#    -q 代表 quiet，減少安裝過程的輸出訊息
#!pip install jieba -q

import jieba # 引入 jieba 函式庫
import string # 用於取得標準英文標點符號
import sys # 用來判斷是否在 Colab 環境

print("jieba 安裝完成並匯入所需函式庫。")

# --- 重複之前的 NLTK 語料庫載入檢查 (確保獨立執行性) ---
# 2. 處理 SSL (同上一步)
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

# 3. 下載 Sinica Treebank (同上一步)
try:
    print("\n檢查 Sinica Treebank 語料庫是否存在...")
    _ = sinica_treebank.sents()
    print("語料庫已存在或載入成功。")
except LookupError:
    print("Sinica Treebank 語料庫未找到，開始下載...")
    nltk.download('sinica_treebank')
    print("語料庫下載完成。")
except Exception as e:
    print(f"載入或下載語料庫時發生錯誤: {e}")
    # 如果下載/載入失敗，則停止執行
    exit()
# --- 語料庫載入檢查結束 ---

print("\n" + "="*30)
print("開始處理 Sinica Treebank 並比較斷詞結果")
print("="*30 + "\n")

# 4. 定義要移除的標點符號集合
#    string.punctuation 包含標準英文標點: '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
#    再加上中文常用的標點符號 (全形與半形)
chinese_punctuation = "，。？！；：、…—-（）《》【】「」『』～·"
punctuation_set = set(string.punctuation + chinese_punctuation)
print(f"將移除的標點符號集合 (部分範例): {list(punctuation_set)[:10]} ... 等共 {len(punctuation_set)} 個")

# 5. 取得 Sinica Treebank 內建斷詞結果
sinica_sents = sinica_treebank.sents()
print(f"語料庫總句子數: {len(sinica_sents)}\n")

print("開始並列比較 NLTK 內建斷詞 與 Jieba 斷詞(已清理) 結果：")
print("-" * 50)

# 6. 迭代處理每個句子
count = 0
for sinica_segmented_sentence in sinica_sents:
    # a. 從 NLTK 的斷詞結果還原原始句子
    original_sentence = "".join(sinica_segmented_sentence)

    # b. 使用 jieba 進行斷詞 (精確模式，返回 list)
    #    jieba.cut 返回一個 generator，jieba.lcut 直接返回 list
    jieba_segmented_words = jieba.lcut(original_sentence)

    # c. 清理 jieba 結果：移除標點符號和空字串
    #    使用列表推導式 (List Comprehension)
    #    條件：該詞不在標點符號集合中，且去除前後空白後不是空字串
    cleaned_jieba_words = [
        word for word in jieba_segmented_words
        if word not in punctuation_set and word.strip()
    ]

    # d. 並列印出比較結果
    #    為了對齊和可讀性，可以稍微格式化輸出
    print(f"原始句子: {original_sentence}")
    print(f"NLTK 斷詞: {sinica_segmented_sentence}")
    print(f"Jieba 清理: {cleaned_jieba_words}")
    print("-" * 20) # 使用分隔線區分每個句子

    count += 1
    # # 如果語料庫很大，可以取消下面幾行的註解，只印出前幾句作為範例
    # if count >= 10:
    #     print(f"\n(僅顯示前 {count} 句比較)")
    #     break

print("-" * 50)
print(f"\n已成功處理並比較 {count} 個句子。")
```

非常好，請幫我承接上面，撰寫一個能夠針對中文進行「詞性標註（POS Tagging）」的 Python 原始碼。

該程式碼的步驟如下所示：

*   請對original_sentence進行「詞性標註」。
*   完成後將語料庫內建的詞性標註結果： nltk.corpus.sinica_treebank.tagged_sents()，與剛剛產生的詞性標註結果並列，一次一句地全數印出。

撰寫此一程式時，請遵守下列限制：

*   將該程式碼撰寫成適合在 Google Colab 環境下執行的樣子。



```python
# -*- coding: utf-8 -*-
# 在 Google Colab 中執行此程式碼 (接續上一個儲存格)

# 1. 確保 jieba 已安裝 (若此儲存格獨立執行)
#    如果接續執行，此行可省略或註解掉
# !pip install jieba -q

import nltk
from nltk.corpus import sinica_treebank
import jieba
import jieba.posseg as pseg # 引入 jieba 的詞性標註模組
import ssl
import sys

print("jieba 及所需函式庫已匯入。")

# --- 重複之前的 NLTK 語料庫載入檢查 (確保獨立執行性) ---
# 2. 處理 SSL (同上一步)
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

# 3. 下載 Sinica Treebank (同上一步)
try:
    print("\n檢查 Sinica Treebank 語料庫是否存在...")
    # 使用 tagged_sents() 來觸發檢查，因為我們接下來會用到它
    _ = sinica_treebank.tagged_sents()
    print("語料庫已存在或載入成功。")
except LookupError:
    print("Sinica Treebank 語料庫未找到，開始下載...")
    nltk.download('sinica_treebank')
    print("語料庫下載完成。")
    # 重新載入以確保可訪問
    _ = sinica_treebank.tagged_sents()
except Exception as e:
    print(f"載入或下載語料庫時發生錯誤: {e}")
    exit()
# --- 語料庫載入檢查結束 ---

print("\n" + "="*30)
print("開始處理 Sinica Treebank 並比較詞性標註 (POS Tagging) 結果")
print("="*30 + "\n")

# 4. 取得 Sinica Treebank 內建的已標註詞性的句子
#    格式為: [[('詞1', '標籤1'), ('詞2', '標籤2'), ...], [句子2], ...]
sinica_tagged_sents = sinica_treebank.tagged_sents()
print(f"語料庫總句子數: {len(sinica_tagged_sents)}\n")

print("開始並列比較 NLTK 內建標註 與 Jieba 標註 結果：")
print("(注意：兩者的斷詞結果和詞性標籤集可能不同)")
print("-" * 60)

# 5. 迭代處理每個已標註的句子
count = 0
for sinica_tagged_sentence in sinica_tagged_sents:
    # a. 從 NLTK 的標註結果中提取詞語，還原原始句子
    #    sinica_tagged_sentence 是一個 list of (word, tag) tuples
    original_words = [word for word, tag in sinica_tagged_sentence]
    original_sentence = "".join(original_words)

    # b. 使用 jieba.posseg 進行斷詞與詞性標註
    #    pseg.lcut() 直接返回一個 list of pair objects (每個 pair 有 .word 和 .flag 屬性)
    jieba_pos_pairs = pseg.lcut(original_sentence)

    # c. 將 jieba 結果轉換成 (詞, 標籤) 的 tuple 列表，方便比較
    jieba_tagged_result = [(pair.word, pair.flag) for pair in jieba_pos_pairs]

    # d. 並列印出比較結果
    print(f"原始句子: {original_sentence}")
    print(f"NLTK 標註: {sinica_tagged_sentence}")
    print(f"Jieba 標註: {jieba_tagged_result}")
    print("-" * 30) # 分隔線

    count += 1
    # 如果語料庫很大，可以取消下面幾行的註解，只印出前幾句作為範例
    if count >= 10:
      print(f"\n(僅顯示前 {count} 句比較)")
      break

print("-" * 60)
print(f"\n已成功處理並比較 {count} 個句子的詞性標註結果。")
print("提醒：NLTK Sinica Treebank 和 Jieba 使用不同的詞性標籤規範。")
```

非常好，請幫我承接上面，撰寫一個能夠針對中文進行「實體名稱辨識（Named Entity Recognition, NER）」的 Python 原始碼。

該程式碼的步驟如下所示：

*   請對original_sentence進行「實體名稱辨識」，並將產生的結果一次一句地全數印出。若該句不包含實體名稱，則應排除該句之列印。

撰寫此一程式時，請遵守下列限制：

*   將該程式碼撰寫成適合在 Google Colab 環境下執行的樣子。



```python
# -*- coding: utf-8 -*-
# 在 Google Colab 中執行此程式碼 (接續上一個儲存格)

# 1. 確保 jieba 已安裝 (若此儲存格獨立執行)
# !pip install jieba -q

import nltk
from nltk.corpus import sinica_treebank
import jieba
import jieba.posseg as pseg # 引入 jieba 的詞性標註模組，用於 NER
import ssl
import sys

print("jieba 及所需函式庫已匯入。")

# --- 重複之前的 NLTK 語料庫載入檢查 (確保獨立執行性) ---
# 2. 處理 SSL (同上一步)
try:
    _create_unverified_https_context = ssl._create_unverified_context
except AttributeError:
    pass
else:
    ssl._create_default_https_context = _create_unverified_https_context

# 3. 下載 Sinica Treebank (同上一步)
try:
    print("\n檢查 Sinica Treebank 語料庫是否存在...")
    # 使用 sents() 即可，因為我們主要需要原始文本
    _ = sinica_treebank.sents()
    print("語料庫已存在或載入成功。")
except LookupError:
    print("Sinica Treebank 語料庫未找到，開始下載...")
    nltk.download('sinica_treebank')
    print("語料庫下載完成。")
    _ = sinica_treebank.sents() # 確保下載後可訪問
except Exception as e:
    print(f"載入或下載語料庫時發生錯誤: {e}")
    exit()
# --- 語料庫載入檢查結束 ---

print("\n" + "="*30)
print("使用 Jieba 詞性標註進行基礎的實體名稱辨識 (NER)")
print("僅列印包含已辨識實體 (人名、地名、機構名) 的句子")
print("="*30 + "\n")

# 4. 定義通常代表命名實體的 Jieba 詞性標籤集合
#    nr: 人名 (Person Name)
#    ns: 地名 (Place Name / Geopolitical Entity - GPE)
#    nt: 机构团体名 (Organization Name)
#    可以根據需要增減其他標籤，例如 nz (其他專有名詞)
entity_tags = {'nr', 'ns', 'nt'}
print(f"將辨識具有以下詞性標籤的詞語作為實體: {entity_tags}\n")

# 5. 取得 Sinica Treebank 的句子（用於還原原始文本）
sinica_sents = sinica_treebank.sents()
print(f"語料庫總句子數: {len(sinica_sents)}\n")

print("開始處理句子並尋找實體名稱：")
print("-" * 60)

# 6. 迭代處理每個句子
count_total = 0
count_printed = 0
for sentence_tokens in sinica_sents:
    count_total += 1
    # a. 從 NLTK 的斷詞結果還原原始句子
    original_sentence = "".join(sentence_tokens)

    # b. 使用 jieba.posseg 進行斷詞與詞性標註
    jieba_pos_pairs = pseg.lcut(original_sentence)

    # c. 篩選出詞性標籤在 entity_tags 中的 (詞語, 標籤) 對
    identified_entities = [
        (pair.word, pair.flag) for pair in jieba_pos_pairs if pair.flag in entity_tags
    ]

    # d. 如果篩選結果不為空，則表示該句包含實體，印出結果
    if identified_entities: # 只有列表不為空時才執行
        print(f"原始句子: {original_sentence}")
        print(f"辨識出的實體 (Jieba: 詞語, 標籤): {identified_entities}")
        print("-" * 30) # 使用分隔線區分每個句子
        count_printed += 1

    # 可選：限制輸出數量，特別是對於大型語料庫
    if count_printed >= 25: # 例如，只印出前 25 個包含實體的句子
      print(f"\n(已達到輸出上限，停止顯示更多結果)")
      break

print("-" * 60)
print(f"\n處理完成。")
print(f"總共處理了 {count_total} 個句子。")
print(f"其中 {count_printed} 個句子被辨識出包含至少一個實體 (人名、地名、機構名) 並已列印。")
```
